```python
#contoh 1
def hitung_margin (harga_jual, harga_modal):
    """Hitung margin keuntungan dalam persen."""
    margin =    (harga_jual - harga_modal) / harga_jual * 100
    return round(margin,2)

print(hitung_margin(20000,10000))
```

## Soal 3 — Filter Data (List of Dicts)

```python
#Konteks: Filter transaksi high-value (> 500rb) dan ambil ringkasan.
#Target: Buat function filter_high_value(transaksi, threshold) yang mengembalikan list ID transaksi dengan nilai transaksi (harga * qty) > threshold.

transaksi = [
    {"id": "TRX1", "produk": "A", "harga": 150000, "qty": 2},
    {"id": "TRX2", "produk": "B", "harga": 500000, "qty": 1},
    {"id": "TRX3", "produk": "C", "harga": 300000, "qty": 3},
    {"id": "TRX4", "produk": "D", "harga": 800000, "qty": 1},
    {"id": "TRX5", "produk": "E", "harga": 250000, "qty": 4},
]

def filter_high_value(data_transaksi, threshold):            # Definisi function untuk memfilter transaksi high-value
    high_value_id = []                                       # buat list kosong untuk menampung ID transaksi yang memenuhi kriteria
    for trx in data_transaksi:                               # kolomp data_transaksi, ambil setiap trx (dictionary) satu per satu 
        total = trx["harga"] * trx['qty']                    # Hitung total transaksi (harga * qty)
        if total > threshold:                                # Jika total transaksi > threshold, tambahkan ID transaksi ke list high_value_id
            high_value_id.append(trx["id"])                  # Tambahkan ID transaksi ke list high_value_id
    return high_value_id                                     # Kembalikan list ID transaksi high-value

high_value_transactions = filter_high_value(transaksi, 500000)
print("ID Transaksi > 500000:", high_value_transactions)

```

## Soal 4 — Agregasi per Kategori (GROUP BY)

```python
#Konteks: Mirip GROUP BY di SQL. Agregasi total revenue per kategori.
#Target: Total revenue per kategori, urutkan dari terbesar.

transaksi = [
    {"kategori": "Elektronik", "revenue": 1200000},
    {"kategori": "Fashion", "revenue": 450000},
    {"kategori": "Elektronik", "revenue": 800000},
    {"kategori": "Makanan", "revenue": 250000},
    {"kategori": "Fashion", "revenue": 700000},
    {"kategori": "Elektronik", "revenue": 1500000},
]

def total_revenue_per_kategori(transaksi):                  # Definisi function untuk menghitung total revenue per kategori
    revenue_per_kategori = {}                               # Buat dictionary kosong untuk menampung total revenue per kategori
    for trx in transaksi:                                   
        kategori = trx["kategori"]
        revenue = trx["revenue"]
        if kategori in revenue_per_kategori:                # Jika kategori sudah ada di revenue_per_kategori, tambahkan revenue
            revenue_per_kategori[kategori] += revenue       # Tambahkan revenue ke kategori yang sudah ada
        else:
            revenue_per_kategori[kategori] = revenue        # Buat entry baru untuk kategori yang belum ada
    return revenue_per_kategori

total_revenue = total_revenue_per_kategori(transaksi)
# Urutkan dari terbesar
sorted_revenue = sorted(total_revenue.items(), reverse=True, key=lambda x: x[1])    # Urutkan berdasarkan revenue (nilai) secara descending
print(f"Total revenue per kategori (urut dari terbesar):")
for x,y in sorted_revenue:
    print(f"- {x:<10}: Rp{y:>10,.0f}")
```

## Soal 5 — Deteksi Nilai Hilang (None / Missing)

```python
## Konteks: Anda menemukan data dengan nilai kosong. Cek dan laporkan.
## Target: Hitung berapa baris yang memiliki minimal 1 nilai None, dan kolom mana yang paling banyak missing.

data = [
    {"nama": "Andi", "usia": 25, "kota": "Jakarta"},
    {"nama": "Budi", "usia": None, "kota": "Bandung"},
    {"nama": "Citra", "usia": 30, "kota": None},
    {"nama": "Dewi", "usia": 28, "kota": "Surabaya"},
    {"nama": "Eka", "usia": None, "kota": None},
]

# Hitung baris dengan nilai None
baris_dengan_none = 0                                   # Inisialisasi counter untuk baris dengan nilai None
for baris in data:                                      # Iterasi setiap baris (dictionary) dalam data
    if any(v is None for v in baris.values()):          # Jika ada nilai None di baris tersebut, tambahkan counter
        baris_dengan_none +=1

# Kolom dengan nilai None terbanyak
kolom_none_count = {}                                   # Inisialisasi dictionary untuk menghitung jumlah None per kolom
for baris in data:                                      # Iterasi setiap baris (dictionary) dalam data
    for k, v in baris.items():                          # Iterasi setiap key-value pair dalam baris
        if v is None:                                   # Jika nilai adalah None, tambahkan ke counter kolom
            if k in kolom_none_count:                   # Jika kolom sudah ada di kolom_none_count, tambahkan 1
                kolom_none_count[k] += 1
            else:                                       # Jika kolom belum ada di kolom_none_count, buat key baru dengan nilai 1
                kolom_none_count[k] = 1                 

print(f"\nJumlah baris dengan minimal 1 None: {baris_dengan_none}")

print("Jumlah None per kolom:" )
for k,v in kolom_none_count.items():
    print(f"- {k:<5}: {v}")          
```

## Soal 6 — Deteksi Duplikat

```python
#Konteks: Anda curiga data hasil join SQL ada duplikasi. Cek dengan Python.
#Target: Identifikasi ID yang duplikat.

transaksi = [
    {"id": "TRX001", "produk": "Laptop"},
    {"id": "TRX002", "produk": "Mouse"},
    {"id": "TRX001", "produk": "Laptop"},   # duplikat
    {"id": "TRX003", "produk": "Keyboard"},
    {"id": "TRX002", "produk": "Mouse"},     # duplikat
]

id_count = {}
for trans in transaksi:
    trx_id = trans["id"]            # Ambil ID transaksi dari dictionary
    if trx_id in id_count:          # Jika ID transaksi sudah ada di id_count, tambahkan 1
        id_count[trx_id] += 1
    else:                           # Jika ID transaksi belum ada di id_count, buat key baru dengan nilai 1
        id_count[trx_id] = 1

print("ID Transaksi yang duplikat:")
for trx_id, count in id_count.items():
    if count > 1:
        print(f"- {trx_id}: {count} kali")
```

```python
#Cara cepat menggunakan get() untuk menghitung frekuensi ID transaksi:
id_count = {}
for trx in transaksi:
    id_count[trx["id"]] = id_count.get(trx["id"], 0) + 1

duplikat = {k: v for k, v in id_count.items() if v > 1}

print(f"Total baris  : {len(transaksi)}")
print(f"Unik ID      : {len(id_count)}")
print(f"Duplikat     : {duplikat}")
```

## Soal 7 — Function Analisis Lengkap

```python
# Konteks: Bungkus semua logika analisis menjadi satu function yang bisa dipakai berulang.
# Target: Function laporan(transaksi) yang mengembalikan dictionary berisi:
# total_revenue, revenue_per_kota (dict), revenue_per_kategori (dict), top_kota (string nama kota dengan revenue tertinggi)

transaksi = [
    {"kota": "Jakarta", "kategori": "Elektronik", "revenue": 1200000},
    {"kota": "Bandung", "kategori": "Fashion", "revenue": 450000},
    {"kota": "Jakarta", "kategori": "Fashion", "revenue": 600000},
    {"kota": "Surabaya", "kategori": "Elektronik", "revenue": 900000},
    {"kota": "Jakarta", "kategori": "Elektronik", "revenue": 800000},
    {"kota": "Bandung", "kategori": "Elektronik", "revenue": 700000},
]

def laporan(transaksi):
    total_revenue = 0
    revenue_per_kota = {}
    revenue_per_kategori = {}
    top_kota = ""

    for trx in transaksi:
        kota = trx["kota"]
        kategori = trx["kategori"]
        revenue = trx["revenue"]
        total_revenue += revenue

        # Menghitung revenue per kota
        if kota in revenue_per_kota:
            revenue_per_kota[kota] += revenue
        else:
            revenue_per_kota[kota] = revenue

        # Menghitung revenue per kategori
        if kategori in revenue_per_kategori:
            revenue_per_kategori[kategori] += revenue
        else:
            revenue_per_kategori[kategori] = revenue

    # Menentukan kota dengan revenue tertinggi
    if revenue_per_kota:
        top_kota = max(revenue_per_kota, key=revenue_per_kota.get)
        return {
            "total_revenue": total_revenue,
            "revenue_per_kota": revenue_per_kota,
            "revenue_per_kategori": revenue_per_kategori,
            "top_kota": top_kota
        }

print("Laporan Transaksi:")
laporan_hasil = laporan(transaksi)
for k, v in laporan_hasil.items():
    print(f"{k:20}: {v}")
```

```python
# Konteks: Simulasi mini pipeline — bersihkan, validasi, agregasi, report. Ini adalah rangkuman semua skill hari ini.
# Target: Filter baris tidak valid (produk kosong / harga bukan angka / qty ≤ 0), Konversi tipe untuk baris valid, Hitung total revenue, Laporkan baris yang dibuang.

raw_data = [
    {"id": "1", "produk": "Laptop",  "harga": "12000000", "qty": "2"},
    {"id": "2", "produk": "Mouse",   "harga": "150000",   "qty": "5"},
    {"id": "3", "produk": "",        "harga": "abc",      "qty": "1"},   # invalid
    {"id": "4", "produk": "Keyboard","harga": "350000",   "qty": "3"},
    {"id": "5", "produk": "Monitor", "harga": "2000000",  "qty": "0"},   # qty 0
]

# Bersihkan dan lakukan validasi
def bersihkan(raw_data):
    valid_data = []
    invalid_rows = [] # List untuk menyimpan baris yang dibuang
    for row in raw_data:
        try:
            harga = int(row["harga"])
            qty = int(row["qty"])
            if row["produk"] and harga > 0 and qty > 0:
                valid_data.append(row)
            else:
                invalid_rows.append(row) # Tambahkan baris yang dibuang ke invalid_rows
        except ValueError:
            invalid_rows.append(row) # Tambahkan baris yang dibuang ke invalid_rows
    return valid_data, invalid_rows

valid_data = bersihkan(raw_data)[0]
invalid_rows = bersihkan(raw_data)[1]

# Hitung total revenue
total_revenue = sum(int(row["harga"]) * int(row["qty"]) for row in valid_data)

# Laporan Akhir
print("Laporan Akhir:")

print(f"Baris valid : {len(valid_data)} / {len(raw_data)}")
print(f"Baris ditolak: {len(invalid_rows)}\n")

print(f"Total revenue    : Rp{total_revenue:,.2f}")

    
```
